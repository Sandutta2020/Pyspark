```python
import pandas as pd
```

```python
df_movies_pd = pd.read_csv("/Volumes/movie_catalog/movie_schema/movie_vol/Movie_details.gzip", compression="gzip")
df_movies= spark.createDataFrame(df_movies_pd)
```

```python
%python
catalog = "movie_catalog"
schema = dbName = db = "movie_schema"
volume_name = "monthwise_rating_data"

spark.sql(f'CREATE CATALOG IF NOT EXISTS `{catalog}`')
spark.sql(f'USE CATALOG `{catalog}`')
spark.sql(f'CREATE SCHEMA IF NOT EXISTS `{catalog}`.`{schema}`')
spark.sql(f'USE SCHEMA `{schema}`')
spark.sql(f'CREATE VOLUME IF NOT EXISTS `{catalog}`.`{schema}`.`{volume_name}`')
volume_folder =  f"/Volumes/{catalog}/{db}/{volume_name}"
```

```python
df_movies.coalesce(1).write.format("json").mode("overwrite").save(volume_folder)
```

```python
df_movies.write.saveAsTable('movie_catalog.movie_schema.Movie_dim')
```
