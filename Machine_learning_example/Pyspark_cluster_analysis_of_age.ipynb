```python
from pyspark.sql import SparkSession
spark =SparkSession.builder.appName("People_data_analysis").getOrCreate()
df_people=spark.read.parquet('/Volumes/movie_catalog/movie_schema/temp_storage/people_data')
```

```python
from pyspark.ml.feature import VectorAssembler,MinMaxScaler
from pyspark.ml.clustering import KMeans
```

```python
features =df_people.select("age","salary")
```

```python
assembler =VectorAssembler(inputCols=["age","salary"],outputCol="features")
```

```python
Feature_vector=assembler.transform(features)
```

```python
scaler =MinMaxScaler(inputCol="features",outputCol="scaled_features")
scaler_model=scaler.fit(Feature_vector)
scaled_data=scaler_model.transform(Feature_vector)
```

```python
kmeans =KMeans(featuresCol="scaled_features",k=3,seed=42)
```

```python
model=kmeans.fit(scaled_data)
clusted_data=model.transform(scaled_data)
clusted_data.show()
```
