```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("DepartureDelayAnalysis").getOrCreate()
```

```python
df_dep_delays = spark\
.read\
.option("inferSchema", "true")\
.option("header", "true")\
.csv("/databricks-datasets/flights/departuredelays.csv")
df_dep_delays.show(5)
```

```python
# Getting the counts of total records
df_dep_delays.count()
```

```python
from pyspark.sql.functions import col
df_dep_delays.orderBy(col("date")).show()
```

```python
# Getting the dates
df_dep_delays[['date']].distinct().count()
```

```python
#Getting necessary packages--------
from pyspark.sql.functions import max,min,count,col,desc
```

```python
#getting the statistics about
df_dep_delays.describe().show()

```

```python
df_dep_delays.corr("delay","distance")
```

```python
df_dep_delays.groupBy("origin","destination").\
     agg(count(col("date")).alias("Flight_count")).show()
```
