```python
import pandas as pd

df =df_dep_delays[[col("date")]].distinct().toPandas()
df["converted_date"]=pd.to_datetime(df['date'],unit='m',origin='unix')
df.head()
```

```python
df_changed_date=spark.createDataFrame(df)
```

```python
df_changed_date.where(col("date") ==1010005).show()
```

```python
df_dep_delays=df_dep_delays.join(df_changed_date,on="date").select("date","delay","distance","origin","destination","converted_date")
df_dep_delays.show(5)
```

```python
#checking based on maximum records based on origin and destination
from pyspark.sql.functions import lit
df_dep_delays.groupBy("origin","destination").agg(count(lit(" ")).alias("flightcount")).orderBy(desc("flightcount")).limit(2).show()
```

```python
#loading the value into variable
df_max_records=df_dep_delays.groupBy("origin","destination").agg(count(lit(" ")).alias("flightcount")).orderBy(desc("flightcount")).first()
```

```python
#checking the values
df_max_records['origin'],df_max_records['destination'],df_max_records['flightcount']
```

```python
df_dep_delays.filter((col("origin")==df_max_records['origin'])&(col("destination")==df_max_records['destination'])).orderBy("date").show()
```

```python
from pyspark.sql.functions import concat
df_dep_delays.filter((col("origin")==df_max_records['origin'])&(col("destination")==df_max_records['destination'])).count()
#    withColumn("fdetail",concat(col(df_dep_delays['origin']),lit("-"),col(df_dep_delays['destination']))).show()
```

```python
df_delay_filtered1=df_dep_delays.filter((col("origin")==df_max_records['origin'])&(col("destination")==df_max_records['destination'])).withColumn("fdetail",concat(col('origin'),lit("--"),col('destination')))
```

```python
df_delay_filtered2=df_dep_delays.filter((col("destination")==df_max_records['origin'])&(col("origin")==df_max_records['destination'])).withColumn("fdetail",concat(col('origin'),lit("--"),col('destination')))
df_delay_filtered2.count()
```
