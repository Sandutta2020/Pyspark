```python
JSON1 =[{"NAME":"SANTANU1","SKILL":"PYTHON"},
        {"NAME":"SANTANU2","SKILL":"JAVA"},
        {"NAME":"SANTANU3","SKILL":"JAVA"}]
```

```python
from pyspark.sql import SparkSession
spark =SparkSession.builder.appName("Dataframe_JSON").getOrCreate()
df = spark.createDataFrame(JSON1)
df.show()
df.printSchema()
```

```python
result=spark.sql("select current_timestamp() as Current_time").collect()
print(result)
current_time =result[0]["Current_time"]
print(current_time)
```

```python
JSON2 =[{"NAME":"SANTANU1","SKILL":"PYTHON","utc_date":current_time},
        {"NAME":"SANTANU2","SKILL":"JAVA","utc_date":current_time},
        {"NAME":"SANTANU3","SKILL":"JAVA","utc_date":current_time}]
```

```python
from pyspark.sql import SparkSession
spark =SparkSession.builder.appName("Dataframe_JSON").getOrCreate()
df_new = spark.createDataFrame(JSON2)
df_new.show(truncate=False)
df_new.printSchema()
```
