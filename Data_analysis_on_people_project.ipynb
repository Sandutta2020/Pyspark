```python
def Return_month_name(month_id):
    if month_id =='01':
        return 'January'
    elif month_id =='02':
        return 'February' 
    elif month_id =='03':
        return 'March'  
    elif month_id =='04':
        return 'April'
    elif month_id =='05':
        return 'May' 
    elif month_id =='06':
        return 'June' 
    elif month_id =='07':
        return 'July'
    elif month_id =='08':
        return 'August' 
    elif month_id =='09':
        return 'September' 
    elif month_id =='10':
        return 'October'
    elif month_id =='11':
        return 'November' 
    elif month_id =='12':
        return 'December'
    else:
        return 'No Match'
         
```

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType
pysp_Return_month_name=udf(Return_month_name,StringType())
```

```python
df_people.withColumn("monthName",pysp_Return_month_name(col('Month'))).show()
```

```python
df_people=df_people.withColumn("monthName",pysp_Return_month_name(col('Month')))
```

```python
df_people.groupby("monthName").pivot("gender").avg("salary").show()
```

```python
from pyspark.sql.functions import avg,mean,count
df_people.groupby("monthName").pivot("gender").agg(avg("salary").alias("Avg_Salary")).show()
```

```python
from pyspark.sql.functions import avg,mean,count,round
df_people.groupby("gender").\
    pivot("monthName",["January","February","March"]).\
        agg(avg("salary").alias("avg_salary"),\
            count("id").alias("cnt")).show()
```

```python
from pyspark.sql.functions import avg,mean,count,round
df_people.groupby("gender").\
    pivot("monthName",["January","February","March"]).\
        agg(round(avg("salary"),2).alias("avg_salary"),\
            count("id").alias("cnt")).show()
```
