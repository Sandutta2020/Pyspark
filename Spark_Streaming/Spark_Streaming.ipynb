```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("Spark_Streaming").getOrCreate()
```

```python
df =spark.readStream.format("cloudFiles")\
    .option("cloudFiles.schemaLocation","/Volumes/main/autoloader/bronze/destination/checkpoint/")\
    .option("cloudFiles.format","csv")\
            .option("cloudFiles.schemaEvolutionMode","rescue")\
                .load("/Volumes/main/autoloader/bronze/raw/")

```

```python
df.writeStream.format("delta")\
    .outputMode("append")\
        .option("checkpointLocation","/Volumes/main/autoloader/bronze/destination/checkpoint/")\
            .trigger(once=True)\
                .start("/Volumes/main/autoloader/bronze/destination/data/")
```

```python
df1=spark.read.format("delta").load("/Volumes/main/autoloader/bronze/destination/data/")
```
